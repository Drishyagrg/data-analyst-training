```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#univariant graphical analysis
-single columns
-possible values either catrgorical or numerical data
1.Categorical data
  a.countplot:count the frequency of category
  b.piechart

#simply,a type of barplot but use only one value
2.Numerical data
    a. boxplot
    b. histplot

# bivariate graphical analysis
-categorical vs categorical
-categorical vs numerical
-numerical vd numerical:
    1. scatter plot
```

```python
sns.countplot(data=df,x='Status',palette='Set3')
#uses palette for changing color
```

```python
#specific color caiyo vney
color=('red','orange')
plt.figure(figsize=(4,5))
sns.countplot(data=df,x='Status',palette=color)
plt.title('Status of Country',fontsize=18,color='blue',fontstyle='italic',fontweight='bold')
plt.xlabel('Status',fontsize=18,color='blue',fontstyle='italic',fontweight='bold')
plt.ylabel('Frequency',fontsize=18,color='blue',fontstyle='italic',fontweight='bold')
plt.savefig('status.jpg')
plt.show()
```

```python
sns.countplot(data=df,x='Country',palette='Set1')
```

```python
#distributions
#normal distributions and formulas
#z-score hru for next cls
```

```python
counts=df['Status'].value_counts().reset_index()
counts
```

```python
plt.pie(counts['count'], labels= counts['Status'],autopct='%1.1f%%', colors=['red','blue'], startangle=90)
plt.title('Piechart of status')
plt.savefig('piechart.png',bbox_inches='tight', dpi=200)
plt.show()
```

```python
counts=df['Country'].value_counts().reset_index()
counts
plt.pie(counts['Country'], labels= counts['count'],autopct='%1.1f%%', startangle=90)
plt.title('Piechart of status')
plt.savefig('piechart1.png',bbox_inches='tight', dpi=200)
plt.show()
```

```python
column=['Life expectancy ', 'Adult Mortality',
       'infant deaths', 'Alcohol', 'percentage expenditure', 'Hepatitis B',
       'Measles ', ' BMI ', 'under-five deaths ', 'Polio', 'Total expenditure',
       'Diphtheria ', ' HIV/AIDS', 'GDP', 'Population',
       ' thinness  1-19 years', ' thinness 5-9 years',
       'Income composition of resources', 'Schooling']


for i in column:
    sns.boxplot(df[i])
    plt.title(f'boxplot of {i}')
    plt.show()
```

```python
# not suitable for boxplot
for i in df.columns:
    if df[i].dtype in ['int64','float64']:
        plt.figure()
        sns.boxplot(data=df, y=i)
    else:
        print('not working')
```

```python
sns.histplot(data=df, x='Adult Mortality',bins=40,kde=True)
```

```python
sns.histplot(data=df, x='under-five deaths ', bins=30, kde=True)
```

```python
for i in column:
    sns.histplot(data=df, x=df[i], bins=35, kde=True, color='red')
    plt.title(f'histplot of {i}')
    plt.savefig(f'image_of_{i}.jpg',bbox_inches='tight')
    plt.show()
```
